```python
# 1. Install dependencies, Ollama, and pyngrok
!sudo apt-get update -y
!sudo apt-get install -y zstd
!curl -fsSL https://ollama.com/install.sh | sh
%pip install -q pyngrok

import os
import threading
import subprocess
import time
from pyngrok import ngrok

# --- ⚙️ CONFIGURATION ---
NGROK_AUTHTOKEN = "3DMdHOb0lfnsNRDq2I3uA1EhK5O_6q8yphm79KgfBEzdss1iW"
MODEL_TO_PULL = "qwen2.5:7b" # Change this to mistral, phi4, or gemma2 as needed
# ------------------------

# Forcefully kill any lingering ngrok processes at the OS level
os.system("pkill -f ngrok")

# 2. Function to start Ollama server in the background
def start_ollama_server():
    # Bind to all interfaces to ensure ngrok can route traffic to it
    os.environ['OLLAMA_HOST'] = '0.0.0.0:11434'
    os.environ['OLLAMA_ORIGINS'] = '*'
    # Start the background process
    subprocess.Popen(["ollama", "serve"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

# Start server thread
print("Starting local Ollama server...")
ollama_thread = threading.Thread(target=start_ollama_server)
ollama_thread.daemon = True
ollama_thread.start()

# Give the server a moment to initialize
time.sleep(5)

# 3. Pull the requested model
print(f"Downloading model '{MODEL_TO_PULL}'... (This may take a few minutes)")
get_ipython().system(f"ollama pull {MODEL_TO_PULL}")

# 4. Expose the local server to the public internet using ngrok
print("\nOpening ngrok tunnel...")

# Authenticate ngrok
ngrok.set_auth_token(NGROK_AUTHTOKEN)

# Ollama runs on port 11434 by default
tunnel = ngrok.connect(11434, "http")

print("\n" + "="*60)
print("🚀 SUCCESS! Your Ollama public API is now live.")
print(f"🔗 Public URL: {tunnel.public_url}")
print("="*60)

print("\n💡 Test your API from your local computer's terminal using:")
print(f"""
curl {tunnel.public_url}/api/generate -d '{{
  "model": "{MODEL_TO_PULL}",
  "prompt": "Explain the concept of an API in one sentence.",
  "stream": false
}}'
""")
```
